```python
import subprocess
from IPython.display import Image, display

def run_pymol_script(script_text, out_png, width=1200, height=900):

    script_path = out_png.replace('.png', '.pml')
    with open(script_path, 'w') as f:
        f.write(script_text)
        f.write(f"\nray {width}, {height}\n")
        f.write(f"png {out_png}, dpi=200\n")
        f.write("quit\n")
    result = subprocess.run(['pymol', '-cq', script_path], capture_output=True, text=True)
    print(result.stdout[-1000:])
    if result.returncode != 0:
        print("STDERR:", result.stderr[-1000:])
    display(Image(out_png))
```

### 1. AMPK heterotrimer (PDB 4RER)


```python
ampk_script = """
fetch 4rer, async=0
hide everything
bg_color white
show cartoon
color skyblue, chain A
color salmon, chain B
color palegreen, chain C
show sticks, resn AMP
color yellow, resn AMP
show sticks, resn STU
color magenta, resn STU
set ray_opaque_background, 1
orient
zoom all, 5
"""
run_pymol_script(ampk_script, "ampk_heterotrimer.png")
```

### 2. mTOR–mLST8 kinase complex (PDB 4JSP)

```python
mtor_script = """
fetch 4jsp, async=0
hide everything
bg_color white
show cartoon
color lightorange, chain A
color slate, chain B
show sticks, resn AGS
color yellow, resn AGS
show spheres, resn MG
color green, resn MG
set sphere_scale, 0.4, resn MG
set ray_opaque_background, 1
orient
zoom all, 5
"""
run_pymol_script(mtor_script, "mtor_kinase_complex.png")
```

### 3. Mitochondrial respiratory Complex I (PDB 5XTD)

```python
complex1_script = """
fetch 5xtd, async=0
hide everything
bg_color white
show cartoon
spectrum chain, rainbow
set ray_opaque_background, 1
orient
zoom all, 5
"""
run_pymol_script(complex1_script, "complex_I_structure.png", width=1000, height=750)
```

### 4. VEGF / Flt-1 (VEGFR1) domain 2 complex (PDB 1FLT)

```python
vegf_script = """
fetch 1flt, async=0
hide everything
bg_color white
show cartoon
color firebrick, chain V or chain W
color deepteal, chain X or chain Y
show sticks, resn CYS and (chain V or chain W)
color orange, resn CYS and (chain V or chain W)
set ray_opaque_background, 1
orient
zoom all, 5
"""
run_pymol_script(vegf_script, "vegf_flt1_complex.png")
```

### 5. Metformin bound to its transporter OCT1 (PDB 8SC4)

```python
oct1_script = """
fetch 8sc4, async=0
hide everything
bg_color white
show cartoon
color wheat, polymer
show sticks, resn MF8
show spheres, resn MF8
set sphere_scale, 0.25, resn MF8
color purple, resn MF8
set ray_opaque_background, 1
orient resn MF8
zoom resn MF8, 8
"""
run_pymol_script(oct1_script, "metformin_oct1.png")
```
