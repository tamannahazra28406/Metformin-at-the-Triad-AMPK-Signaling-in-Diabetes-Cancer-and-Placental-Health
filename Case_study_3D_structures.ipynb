### 1. AMPK heterotrimer (PDB 4RER)
Alpha (kinase, blue), beta (scaffold, salmon), and gamma (AMP-sensing, green) subunits. AMP (yellow)
and staurosporine (magenta, occupying the kinase active site in this crystal form) shown as sticks.
**Click and drag to rotate, scroll to zoom.**


```python
view = py3Dmol.view(query='pdb:4RER', width=800, height=550)
view.setStyle({'cartoon': {'colorscheme': {'prop':'chain', 'map': {'A':'skyblue','B':'salmon','C':'lightgreen'}}}})
view.addStyle({'resn': 'AMP'}, {'stick': {'colorscheme':'yellowCarbon'}})
view.addStyle({'resn': 'STU'}, {'stick': {'colorscheme':'magentaCarbon'}})
view.zoomTo()
view.spin(True)
view.show()
```

### 2. mTOR–mLST8 kinase complex (PDB 4JSP)
mTOR kinase domain (orange) with its obligate partner mLST8 (light blue). ATP-γ-S (yellow) and
Mg2+ ions (green spheres) mark the active site AMPK signaling restricts access to.

```python
view = py3Dmol.view(query='pdb:4JSP', width=800, height=550)
view.setStyle({'cartoon': {'colorscheme': {'prop':'chain', 'map': {'A':'orange','B':'lightblue'}}}})
view.addStyle({'resn': 'AGS'}, {'stick': {'colorscheme':'yellowCarbon'}})
view.addStyle({'resn': 'MG'}, {'sphere': {'color':'green', 'scale':0.4}})
view.zoomTo()
view.spin(True)
view.show()
```

### 3. Mitochondrial respiratory Complex I (PDB 5XTD)
Where the whole mechanism begins: metformin's mild inhibition here raises the cellular AMP:ATP
ratio. Spectrum-colored by chain since this is a genuinely large, ~45-subunit assembly — the
L-shaped membrane-arm / matrix-arm architecture is visible once it loads (may take a few extra
seconds given the structure's size).

```python
view = py3Dmol.view(query='pdb:5XTD', width=800, height=550)
view.setStyle({'cartoon': {'color':'spectrum'}})
view.zoomTo()
view.spin(True)
view.show()
```

### 4. VEGF / Flt-1 (VEGFR1) domain 2 complex (PDB 1FLT)
VEGF homodimer (red, chains V/W) bound to Flt-1 receptor domain 2 (teal, chains X/Y) — the binding
interface sFlt-1 hijacks as a soluble decoy receptor in preeclampsia.

```python
view = py3Dmol.view(query='pdb:1FLT', width=800, height=550)
view.setStyle({'chain': 'V'}, {'cartoon': {'color':'firebrick'}})
view.setStyle({'chain': 'W'}, {'cartoon': {'color':'firebrick'}})
view.setStyle({'chain': 'X'}, {'cartoon': {'color':'teal'}})
view.setStyle({'chain': 'Y'}, {'cartoon': {'color':'teal'}})
view.zoomTo()
view.spin(True)
view.show()
```

### 5. Metformin bound to its transporter OCT1 (PDB 8SC4)
The actual drug molecule (purple, ligand code `MF8`) captured mid-transport into the cell — the
step that has to happen before any AMPK activation occurs at all. Zoomed to the binding pocket.

```python
view = py3Dmol.view(query='pdb:8SC4', width=800, height=550)
view.setStyle({'cartoon': {'color':'wheat'}})
view.addStyle({'resn': 'MF8'}, {'stick': {'colorscheme':'purpleCarbon'}, 'sphere': {'scale':0.25, 'color':'purple'}})
view.zoomTo({'resn':'MF8'})
view.show()
```
